## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The model predicts whether a content item is declining in search impressions. The label is derived from the data: `is_declining_label = (trend_direction == 'down')`. This is an observed outcome from the raw trend label, not a handcrafted proxy rule.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
from pathlib import Path

path = Path('..') / '..' / 'data' / 'raw' / 'content_refresh_anonymized.csv'
df = pd.read_csv(path)
label_counts = df['trend_direction'].value_counts()
print(label_counts)
print('\nDecline label rows:', (df['trend_direction'] == 'down').sum())
print('Decline label rate:', (df['trend_direction'] == 'down').mean())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule cannot capture the many signal interactions in this task. Decline depends on traffic volume, update age, keyword context, page engagement, and time-series drift together, so the best approach is a learned score that blends these features rather than a single threshold.

```python
# This cell is for CODE (numbers, a query, a check).
import pandas as pd
from pathlib import Path

path = Path('..') / '..' / 'data' / 'raw' / 'content_refresh_anonymized.csv'
df = pd.read_csv(path)
print('Overall decline rate:', (df['trend_direction'] == 'down').mean())
print('\nDecline rate by content type:')
print(df.groupby('content_type')['trend_direction'].apply(lambda x: (x == 'down').mean()))
print('\nDecline rate by age tier:')
print(df.groupby('age_tier')['trend_direction'].apply(lambda x: (x == 'down').mean()))
```
