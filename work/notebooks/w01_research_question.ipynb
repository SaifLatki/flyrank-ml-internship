## 1. My lane (or freestyle) and why

I’m choosing the refresh-priority lane: identify pages that look likely to be declining so an editorial or SEO team can review them first. This feels useful because it is actionable, explainable, and close to the starter dataset’s decline label, which makes it a strong first problem to study over the next 7 weeks.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd
import os

# Load the starter dataset from the repo root
if not os.path.exists('data/raw/content_refresh_anonymized.csv'):
    os.chdir('..')

df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
print('Rows:', len(df))
print('Clients:', df['client_id'].nunique())
print('Declining rate:', round(df['trend_direction'].str.lower().eq('down').mean(), 3))
```

## 2. The question: decision, action, cost of a wrong call

The decision is whether to put a page into a "review now" queue for refresh or content updates. The action belongs to editorial and SEO teams, who can spend limited time on the highest-value pages. A wrong call is costly because it wastes review capacity on pages that are not actually declining, but the cost is still smaller than missing a page that is genuinely losing visibility.

```python
# This cell is for CODE (numbers, a query, a check).
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# A simple check that the label is present and not too rare
label = df['trend_direction'].str.lower().eq('down').astype(int)
print('Positive label count:', int(label.sum()))
print('Positive rate:', round(label.mean(), 3))

# A concrete operational view: if we review the top 50 pages, how many should be declining?
# This is a proxy for the value of ranking well.

# Placeholder ranking by recency and visibility for illustration
score = (df['days_since_last_update'] >= 180).astype(int) * (df['impressions_90d'] >= 500).astype(int)
ranked = df.assign(score=score).sort_values('score', ascending=False)
print('Top 50 expected positives (proxy):', int(ranked.head(50)['trend_direction'].str.lower().eq('down').sum()))
```
